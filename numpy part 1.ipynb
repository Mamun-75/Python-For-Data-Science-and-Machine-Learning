```python
##  random numbers !!!! most important
np.random.rand(5)  # this is going to create an arry of the given shape we pass in
                    ## and its going to populate with random samples from a uniform distribution over 0 to 1
```

```python
np.random.rand(5,5) ## for 2D
```

```python
np.random.randn(4,4)
```

```python
## A uniform distribution has a flat, rectangular shape where every value in a set range has an equal chance of happening. A normal distribution has a symmetric, bell-shaped curve where values near the middle mean are most common, and values far from the center are rare
```

```python
## rand int
np.random.randint(1,100) # here , 1 is inclusive and 100 is exclusive,"Inclusive" (low) means the starting value can be included in the results.
                         ## and "Exclusive" (high) means the ending value cannot be included in the results.
```

```python
# we can pass a third argument  = number of random number,,,,
#inclusive is the low end, exclusive in the high end
np.random.randint(1,100,10)
```
