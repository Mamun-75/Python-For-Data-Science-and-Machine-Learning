```python
import seaborn as sns
import matplotlib.pyplot as plt
iris = sns.load_dataset("iris")
iris.head()
```

```python
## pairplot is simpler and show us the information that we want,
## pairgrid gives a little more customization  ability

g = sns.PairGrid(iris)
g.map_diag(sns.histplot)
g.map_upper(plt.scatter)
g.map_lower(sns.kdeplot)
```

```python
 ## Facetgrid
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
g= sns.FacetGrid(data = tips, col = 'time', row = 'smoker')
g.map(sns.histplot, 'total_bill')
```

```python
g= sns.FacetGrid(data = tips, col = 'time', row = 'smoker')
g.map(plt.scatter, 'total_bill','tip')  ## if the plot  need more than one arguments, we can pass them separated by comma
```
