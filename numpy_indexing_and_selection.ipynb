```python
arr = np.arange(0,11)
```

```python
arr
```

```python
slice_of_arr = (arr[0:6])
```

```python
slice_of_arr
```

```python
slice_of_arr[:]
```

```python
slice_of_arr[:]=99
```

```python
slice_of_arr
```

```python
arr
```

```python
## so slice_of_arr is not a copy of arr,
## its just a reference of the original arr,
## thats why when we broadcast 99 in slice_of_arr , it actually changed the original arr
```

```python
# so if we want to copy an array,
arr_copy = arr.copy()
```

```python
arr
```

```python
arr_copy[:] = 100
```

```python
arr_copy
```

```python
arr  ## now it hasn't changed
```

```python
## for 2D array
```

```python
import numpy as np
```

```python
arr_2d = np.array([[5,10,15],[20,25,30],[35,40,45]])
```

```python
arr_2d
```

```python
## there are 2 general formats for grabbing an element from a 2D array
# 1. double bracket format
# 2. single bracket format
```

```python
arr_2d[1][2]
```

```python
arr_2d[1,2] # recommended
```

```python
## if we want to grab a section instead of a single element
arr_2d
```

```python
arr_2d[:2, 1:]   ## [ rows, cols]
```

```python
arr_2d[:2]
```

```python
arr_2d[1:, 1:]
```

```python
## lets move on to the conditional selection which we will use more in this series
```

```python
arr = np.arange(1,11)
```

```python
arr
```

```python
bool_arr = arr > 5   ## comparision operator here actually return an array of boolean value
```

```python
bool_arr
```

```python
arr[bool_arr]   ## this return value only where boolean array is true
```

```python
## we will use this syntax all the time
arr = np.arange(1,11)

arr[arr>5]
```

```python
arr[arr<3]  ## inside the bracket we actually passing boolean values here
```

```python
## practice
arr_2d = np.arange(50).reshape(5,10)
```

```python
arr_2d
```

```python
## lets grab 13,14 here
arr_2d[1,3:5]
```
