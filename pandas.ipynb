Group By
group together rows based off of a column and perform an aggregate function on them

```python
import numpy as np
import pandas as pd
```

```python
data = {'Company': ['GOOG','GOOG','MSFT','MSFT','FB','FB'],
        'person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
        'Sales':[200,120,340,124,243,350]}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
byComp = df.groupby('Company')
```

```python
byComp['Sales'].mean()
```

```python
byComp.sum()
```

```python
byComp['Sales'].std()
```

```python
byComp.sum().loc['FB']
```

```python
df.groupby('Company').count()
```

```python
df.groupby('Company').max()
```

```python
df.groupby('Company').min()
```

```python
df.groupby('Company').describe().transpose()
```

```python
df.groupby('Company').describe().transpose()['FB']
```

Merging


```python

## the merge function allows you to merge dataframes together using a similar logic as merging sql tables together for example:
## instead of gluing the dataframe pd.merge see where the values match up on columns
## by default we get inner merge
```

```python
pd.merge(left,right,how = 'inner', on = 'key')   ## merging on key column
```

```python
left = pd.DataFrame ({'key1':['k0','k0', 'k1','k1'],
                      'key2':['k0','k1','k0','k1'],
                      'A':['A0','A1','A2','A3'],
                      'B':['B0','B1','B2','B3']})
right = pd.DataFrame ({'key1':['k0','k1', 'k1','k2'],
                      'key2':['k0','k0','k0','k0'],
                       'C': ['C0','C1','C2','C3'],
                       'D': ['D0','D1','D2','D3']})
```

```python
left
```

```python
right
```

```python
pd.merge(left,right, on = ['key1','key2'])
```

```python
pd.merge(left,right, how = 'outer', on = ['key1','key2'])
```

```python
pd.merge(left, right, how = 'right', on = ['key1','key2'])  ## study details about how
```

```python
pd.merge(left, right, how = 'left', on = ['key1','key2'] )  ## study details about how
```

joining

```python
## joining is a convenient method for combining rhe columns of two potentially differently indexed dataframes into a single result DataFrame
```

```python
left = pd.DataFrame({'A':['A0','A1','A2'],
                    'B':['B0','B1','B2']},
                    index = ['K0','K1','k2'])
left = pd.DataFrame({'C':['C0','C2','C3'],
                    'D':['D0','D2','D3']},
                    index = ['K0','K2','K3'])
```

```python
left.join(right, lsuffix='_left', rsuffix='_right')
```

```python
left.join(right, how = 'outer', lsuffix='_left', rsuffix='_right')
```
