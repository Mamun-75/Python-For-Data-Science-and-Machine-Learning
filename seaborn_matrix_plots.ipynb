```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
flights = sns.load_dataset('flights')
tips.head()
```

```python
flights.head()
```

```python
### heatmap (primary level matrix plot)
```

```python
## we need a matrix form of data to call heatmap on it
```

```python
tc = tips.corr(numeric_only=True)  ## this is a matrix form
```

```python
tc
```

```python
sns.heatmap(tc)
"""
heatmap actually colors in those value based on some
some sort of gradient scale.
if you have a very large correlation data frame or some other matrix form data you can
easily tell what the values are relative to each other using a heat map.
this is the main primary use of a matrix plot
"""
```

```python
sns.heatmap(tc, annot = True, cmap = 'coolwarm' ) # annotation shows acutal value with colors, cmap shows color options
```

```python
fp = flights.pivot_table(index = 'month', columns = 'year', values = 'passengers')  # we can create matrix form using pivot table quite easily
```

```python
fp
```

```python
sns.heatmap(fp, cmap = 'coolwarm', linecolor='white', linewidths=1) # we can specify linecolor, linewidth if we want to separate each blocks
```

```python
 ## clustermap
 # it plots a matrix as a color-coded heat map while reordering and grouping similar rows and columns using hierarchical clustering
```

```python
sns.clustermap(fp,cmap='coolwarm', standard_scale=1)
## it looks like nov, jan, feb is kind of winter months are much closer to low passengers
## and high passengers count is mainly in these(august, july) summer months
## a cluster map will cluster things in an order so that similar groups are close to each other,
```
