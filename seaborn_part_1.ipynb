dist plot allows us to visualize the distribution (essentially a histogram) of a univariate set of observations , univariate is another way of saying just one variable.


```python
sns.displot(tips['total_bill'],kde = False, bins = 40) ## kde  means kernal density estimation
## now we essentially have a histogram which is essentially just a distribution of where total bill lies
## we can define number of bins as well
## this dist plot shows most of the bills happen somewhere between 10 and 20 $  and begin to fade away as we go higher and higher
```

joint plot, allows us to match up two dist plot  for bi-variate data, meaning we can essentially combine two distribution plots, bivariate is just two variable

```python
sns.jointplot(x ='total_bill', y = 'tip', data = tips, kind = 'hex')  ## kind = kde, reg, hex,,, each has different significance
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,kind='reg')
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,kind='kde')
```

```python
### rugplot  ## rugplot

## rugplots are actually a very simple concept, they just draw a dash mark for every point on a univariate distribution. They are the building block of a KDE plot:
##A rug plot in Seaborn visualizes the exact distribution of a variable by drawing small tick marks along an axis for every individual data point.



sns.rugplot(tips['total_bill'])
```

```python
sns.displot(tips['total_bill'], kde = True)
```

```python
#kde plot stands for kernal density estimation plot
## KDE (Kernel Density Estimate) is a statistical technique used to visualize the shape of a data distribution
## by creating a smooth, continuous curve over your data points, acting essentially as a smoothed-out version of a histogram that shows where values are most densely clustered.
"""
A KDE (Kernel Density Estimate) plot is a smooth, continuous curve
that visualizes the probability distribution of your data.
Think of it as a fluid version of a histogram—
it shows exactly where your data values are densely clustered (high peaks)
and where they are sparse (low valleys) without sorting them into rigid, arbitrary bins

"""

import seaborn as sns
import matplotlib.pyplot as plt

# 1. Load the built-in tips dataset
tips = sns.load_dataset('tips')

# 2. Draw the KDE plot for the 'total_bill' column
# Adding fill=True shades the area under the curve for better visibility
sns.kdeplot(data=tips, x='total_bill', fill=True, color='purple')

# 3. Add a title and display
plt.title('KDE Plot of Total Bills')
plt.show()
```

A KDE (Kernel Density Estimate) plot is a smooth, continuous curve that visualizes the distribution of your data, acting like a fluid version of a histogram that highlights exactly where values are most densely clustered without sorting them into arbitrary bins. To calculate it, the algorithm places a small, standardized shape—usually a mini bell curve called a "kernel"—directly centered on every single individual data point in your dataset. The width of these tiny curves is determined by a setting called the bandwidth. Finally, the algorithm mathematically adds up the overlapping heights of all these individual kernels at every single point along the axis, drawing one seamless line over the total summed height to reveal the true underlying shape of your data.
