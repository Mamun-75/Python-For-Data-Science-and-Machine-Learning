Categorical plot


```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

tips = sns.load_dataset('tips')
tips.head()
```

```python
## here we are mainly concern about categorical column such as sex, smoker etc. categorical column
```

```python
## most basic categorical plot _ barplot
## here x is a categorical column , y is a numerical column
sns.barplot(  x = 'sex', y ='total_bill', data = tips  )  ## showing average total bill for male and female
```

```python
import numpy as np

sns.barplot(  x = 'sex', y ='total_bill', data = tips , estimator=np.std ) # estimator is an aggregate function, by default it is mean, we can change if we want
## here we used standard deviation as estimator and we found that male customers hve more variance than female
```

```python
## in countplot , estimator explicitly count the number of occurances
sns.countplot(x = 'sex', data = tips)
## here it shows exact number of female and male . this only take a single categorical column
## this dataset is showing there are more male than female here
```

```python
## box plot
# it shows the distribution of quantitative data in a way that hopefully facilitates comparisons between variables
```

```python
sns.boxplot(x = 'day', y = 'total_bill', data = tips) # it should show distribution of total bill per day
```

```python
sns.boxplot(x = 'day', y = 'total_bill', data = tips, hue = 'smoker')  ## hue allows us to split data based on a categorical column, which is 'smoker' here, along with a day (which is our main categorical column here)
# is a true power behind seaborn, it quickly allow us to add this hue indicator which essentially gives us a whole another layer of data very quickly and smoothly
# here we not only can explain the characteristics of total bill per day, but also we can explain in which days either smoker have larger bills or  non smokers,
# we can say from this boxplot , fridays tends to have a larger bill if they are not a smoker versus if they are a smoker
```

```python
## violin plot
sns.violinplot(x = 'day', y = 'total_bill', data = tips)
```

```python
sns.violinplot(x = 'day', y = 'total_bill', data = tips, hue = 'sex', split = True)
```

```python
## we should know our audience , if we dealing with some more technical data based people who are data scientists themselves and have seen violin plots before, these sorts of plots are a great choice(violin plot), however for dealing with more kind of CEO executive types management people , may be just a simple box plot which is more popular may be a correct choice , unless you find something very interesting in violin plot
```

```python
## strip plot  -- its just a scatterplot , based off of the catefory
sns.stripplot(x = 'day', y = 'total_bill', data = tips, jitter = True, hue = 'sex', )
```

```python
## swarm plot -- the idea is to combine a strip plot and violin plot
# it shows the same distribution except it's showing all the points themselves
# one drawback sometimes they don't actually scale that well to very large numbers . we advice not to use swarn plots for very large data sets
```

```python
sns.swarmplot(x = 'day', y = 'total_bill', data = tips, hue = 'sex')
```

```python
# lets combine swarm plot and violin plot
sns.violinplot(x = 'day', y = 'total_bill', data = tips)
sns.swarmplot(x = 'day', y = 'total_bill', data = tips, color = 'black' )
"""
here we can see every single points and the corresponding KDE plot on top
of it, now, there is a lot of information here so we need to be careful of this
This is probably more useful not for presentation but just for exploratory data analysis
hopefully this kind of plot help you to immediately understand your data, but no so much
on the side of explaining to an executive ( box plot, bar plot or count plot will work at that case)
"""
```

```python
# catplot is a general form of all these plot, kind argument describe the plot itself, previously known as factor plot
sns.catplot(x = 'day', y = 'total_bill', data = tips, kind = 'bar')
```

```python
sns.catplot(x = 'day', y = 'total_bill', data = tips, kind = 'violin')
```
