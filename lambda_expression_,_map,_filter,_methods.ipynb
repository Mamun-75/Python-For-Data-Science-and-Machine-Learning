```python
## lambda expression , map, filter, methods
```

```python
## recalling function
def times2( var):
  return var*2
```

```python
times2(5)
```

```python
# map()
```

```python
seq = [1,2,3,4,5] # this is a list
```

```python
list(map(times2, seq))   # this times2 function mapped to every elements in that list( seq)
```

```python
## lambda expression
```

```python
t = lambda var : var*2
```

```python
t(6)
```

```python
list(map(lambda num: num*3, seq)) # i dont need to define the function formally , we can use lambda expresiion instead
```

```python
## practice: lets we have a list, we want to multipy its elements by 4
val = [1,2,3,4,5]

list(map(lambda num: num*4, val))
```

```python
list(map(lambda a : a*4, val))
```

```python
##filter function
## a filter must be using some sort of function or lambda expression that returns  a boolean value
num = [1,2,3,4,5,6,7]
```

```python
odd = list(filter(lambda a: a % 2 ==1, num))   #lets filter out odd number from the seq list
```

```python
odd
```

```python
even  = list(filter(lambda a : a%2 == 0, num))
```

```python
even
```

```python
x = [(1,2), (3,4),(5,6)]
```

```python
x[0][1] # we can grab element of tuple inside of the list
```

```python
# tuple unpacking allows us to unpack these tuples
for (a,b) in x:
  print(a)
```

```python
for a,b in x:
  print(a)
  print(b)
```
