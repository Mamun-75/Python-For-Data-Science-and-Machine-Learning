```python
import matplotlib.pyplot as plt
```

```python
## lets make something using two numpy arrays
```

```python
import numpy as np
```

```python
x = np.linspace(0, 5, 11)
y = x ** 2
```

```python
x
```

```python
y
```

```python
 ## functional method
```

```python
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('title')
```

```python
## multiple plot on same canvas
```

```python
plt.subplot(1, 2, 1)
plt.plot(x, y, 'r')

plt.subplot(1, 2, 2)
plt.plot(y, x, 'b')
```

```python
## object oriented method
```

```python
fig = plt.figure() ##  this figure object we can think of it as just kind of an imaginary blank canvas.
## we can add a set of axes to this canvas

axes  = fig.add_axes([0.1,0.1,0.8, 0.8])   ## axes use a list .  here list take 4 arguments. left, bottom, width and height argument
axes.plot(x,y)
axes.set_xlabel(  'x'  )
axes.set_ylabel(  'y'  )
axes.set_title(  'title')
```

```python
fig = plt.figure()
axes1 = fig.add_axes([0.1,0.1,0.8, 0.8])
axes2 = fig.add_axes([0.2,0.6,0.3, 0.2])

axes1.plot(x,y)
axes1.set_xlabel(  'x'  )
axes1.set_ylabel(  'y'  )
axes1.set_title(  'larger plot')

axes2.plot(y,x)
axes2.set_xlabel(  'y'  )
axes2.set_ylabel(  'x'  )
axes2.set_title(  'smaller plot')
```

```python
 ## how to create subplot using object oriented method
```

```python
fig, axes = plt.subplots( nrows=3, ncols=3)
plt.tight_layout()  ## fix overlapping problem
```

```python
fig, axes = plt.subplots( nrows=1, ncols=2) #axes object is a list of matplotlib axes
plt.tight_layout()  ## fix overlapping problem
```

```python
axes
```

```python
fig, axes = plt.subplots( nrows=1, ncols=2)
plt.tight_layout()  ## fix overlapping problem
for current_ax in axes:
    current_ax.plot(x,y)
```

```python
fig, axes = plt.subplots( nrows=1, ncols=2)
plt.tight_layout()  ## fix overlapping problem
axes[0].plot(x,y)
```

```python
fig, axes = plt.subplots( nrows=1, ncols=2)  ## tuple unpacking here,, object oriented method
plt.tight_layout()  ## fix overlapping problem
axes[1].plot(x,y)
```

```python
## figure size , aspect ratio and dpi
```

```python
fig = plt.figure(figsize=(8,2))
ax = fig.add_axes([0,0,1,1])
ax.plot(x,y)
plt.tight_layout()
```

```python
fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(8,2))
plt.tight_layout()
axes[0].plot(x,y)
axes[1].plot(y,x)
plt.tight_layout()
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot(x,x**2, label = ' X squared')
ax.plot(x, x**3, label =  'X cubed')
ax.legend(loc = 0) ## this clarify meaning of each curve, and label is mandatory  for using this. sometimes ax.legend() may overlap the actual curve, so we need to fix postion for legend, thats why we use loc, we have multiple values for loc, such as 0,1,2,3,4,5 etc. check documentation for legend for more, for best position we use loc = 0
```

```python
### Legends allow us to  use label text to actually clarify what plot is what plot !
## if values 1 to 10 for loc argument dont satisfy our desired positin , we can use loc = (x_coordinate value, y_coordinate value)
```

```python
## setting colors
```

```python
from matplotlib.lines import lineStyles
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot(x,y, color = 'purple', linewidth = 2, alpha = 0.5, linestyle = '-', marker = 'o', markersize = 20, markerfacecolor = 'Yellow', markeredgewidth = 3, markeredgecolor = 'green')

## we can use rgb hex code as color as well , e.g #FF8C00, linewidth can also be customized for thiner or thicker line,
# alpha argument specify how transparent the line is
# we dont have to write linewidth , lw will work too
## we can define linestyle , in short ls will work too,  supported values are '-', '--', '-.', ':', 'None', ' ', '', 'solid', 'dashed', 'dashdot', 'dotted'
# we can use marker, 'o', '+' etc. see documentation for more ('.', ',', 'o', 'v', '^', '<', '>', '1', '2', '3', '4', '8', 's', 'p', 'P', '*', 'h', 'H', '+', 'x', 'X', 'D', 'd', '|', '_')
## explore different combinations among these arguments
```

```python
fig = plt.figure()
ax  = fig.add_axes([0,0,1,1])
ax.plot(x,y,color = 'purple',lw  = 2, ls = '--')
ax.set_xlim([0,1]) #lower bound and upper bound
ax.set_ylim([0,2])
```
