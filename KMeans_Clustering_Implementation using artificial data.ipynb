Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

Lets make some artificial data to work with, Here, we will learn how to create artificial dataset using make_blobs mainly for clustering algorithm groups data


```python
from sklearn.datasets import make_blobs
```

```python
data = make_blobs(n_samples=200, n_features=2, centers=4, cluster_std=1.8, random_state  = 101 )
# n_samples means we are taking 200 data points
# n_features = 2 means , only two feature
# center = 4 means we are making 4 blobs,
# cluster_std means How tightly packed or loosely spread out each group/blob is
```

Its really fun, we can play with artificial data!  Lets work with KMeans algorithm

```python
data[1]  # these are correct labels for our created data
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=4)
```

```python
kmeans.fit(data[0])
```

```python
kmeans.cluster_centers_
```

```python
kmeans.labels_   ## these are predicted labels
```

```python
fig, (ax1, ax2) = plt.subplots(1,2, sharey=True, figsize=(10,6)) # i am creating a subplot of 1 row and 2 columns, and using tuple unpacking to grab these two set of axes

ax1.set_title('K Means')
ax1.scatter(data[0][:,0], data[0][:,1], c=kmeans.labels_, cmap='rainbow')

ax2.set_title('Original')
ax2.scatter(data[0][:,0], data[0][:,1], c=data[1], cmap = 'rainbow')
```

K means algorithm is unsupervised, so we dont actually have labeled data, but here we used labeled data to show how K Means actually work


Lets learn how to find optimal K value


```python
inertias = []
```

```python
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=101)

    kmeans.fit(data[0])

    inertias.append(kmeans.inertia_)
```

```python
kmeans.inertia_
```

```python
plt.plot(range(1, 11), inertias, marker='o')
plt.title('The Elbow Method')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia (Total Distance)')
plt.show()
```

```python
## THE ELBOW METHOD INTUITION:
# We look for the "elbow" or the sharp bend in the graph to find the optimal 'K'.
# We do NOT choose the absolute minimum point (like K=10 or K=N).
# If we keep increasing K, inertia (total distance) will naturally always drop,
# but it leads to diminishing returns and overly complex, meaningless clusters.
# The "elbow" (e.g., K=4) represents the "sweet spot" — the perfect balance
# where we get distinct, meaningful clusters without unnecessary computational cost
```
